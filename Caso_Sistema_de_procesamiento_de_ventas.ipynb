```python
ventas = [    {"producto": "Laptop", "categoria": "Tecnologia", "cantidad": 2, "precio": 650},    {"producto": "Mouse", "categoria": "Tecnologia", "cantidad": 5, "precio": 20},    {"producto": "Escritorio", "categoria": "Muebles", "cantidad": 1, "precio": 300},    {"producto": "Teclado", "categoria": "Tecnologia", "cantidad": 3, "precio": 45},    {"producto": "Silla", "categoria": "Muebles", "cantidad": 4, "precio": 120},    {"producto": "Monitor", "categoria": "Tecnologia", "cantidad": 2, "precio": 250}]
```

```python

# Programación imperativa/procedimental
print("Programación imperativa/procedimental\n")
```

```python
totalVentas = 0
for venta in ventas:
    subTotal = venta["cantidad"] * venta["precio"]
    totalVentas += subTotal
print(f"Total acumulado de las ventas: {totalVentas}")
```

```python

totalUnidades = 0
for venta in ventas:
  totalUnidades += venta["cantidad"]
print(f"Total de unidades vendidas: {totalUnidades}")
```

```python
ventaMayor = 0
nombreProducto = ""
for venta in ventas:
  subTotal = venta["cantidad"] * venta["precio"]
  if subTotal > ventaMayor:
    ventaMayor = subTotal
    nombreProducto = venta["producto"]
print(f"Venta más alta: {nombreProducto} con un total de {ventaMayor}")
```

```python
totalVentasTecnologia = 0
for venta in ventas:
  if venta["categoria"] == "Tecnologia":
    subTotal = venta["cantidad"] * venta["precio"]
    totalVentasTecnologia += subTotal
print(f"Total acumulado de las ventas de tecnología: {totalVentasTecnologia}")
```
