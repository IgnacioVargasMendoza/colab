```python
ventas = [    {"producto": "Laptop", "categoria": "Tecnologia", "cantidad": 2, "precio": 650},    {"producto": "Mouse", "categoria": "Tecnologia", "cantidad": 5, "precio": 20},    {"producto": "Escritorio", "categoria": "Muebles", "cantidad": 1, "precio": 300},    {"producto": "Teclado", "categoria": "Tecnologia", "cantidad": 3, "precio": 45},    {"producto": "Silla", "categoria": "Muebles", "cantidad": 4, "precio": 120},    {"producto": "Monitor", "categoria": "Tecnologia", "cantidad": 2, "precio": 250}]
```

```python
print("PARTE I – PROGRAMACIÓN IMPERATIVA/PROCEDIMENTAL")

totalVentas = 0
totalUnidades = 0
ventaMayor = 0
nombreProducto = ""
totalVentasTecnologia = 0

#Ejercicio 1
for venta in ventas:
    subTotal = venta["cantidad"] * venta["precio"]
    totalVentas += subTotal
print(f"Total acumulado de las ventas: {totalVentas}")
#Ejercicio 2
for venta in ventas:
  totalUnidades += venta["cantidad"]
print(f"Total de unidades vendidas: {totalUnidades}")
#Ejercicio 3
for venta in ventas:
  subTotal = venta["cantidad"] * venta["precio"]
  if subTotal > ventaMayor:
    ventaMayor = subTotal
    nombreProducto = venta["producto"]
print(f"Venta más alta: {nombreProducto} con un total de {ventaMayor}")
#Ejercicio 4
for venta in ventas:
  if venta["categoria"] == "Tecnologia":
    subTotal = venta["cantidad"] * venta["precio"]
    totalVentasTecnologia += subTotal
print(f"Total acumulado de las ventas de tecnología: {totalVentasTecnologia}")
```

```python
print("PARTE II – PROGRAMACIÓN FUNCIONAL")
# Ejercicio 1
montos_ventas = list(map(lambda v: v["cantidad"] * v["precio"], ventas))
print(f"Monto de cada venta: {montos_ventas}")

# Ejercicio 2
total_ventas_funcional = sum(montos_ventas)
print(f"Total acumulado de las ventas: {total_ventas_funcional}")

# Ejercicio 3
ventas_tecnologia = list(filter(lambda v: v["categoria"] == "Tecnologia", ventas))
print(f"Ventas de Tecnología: {ventas_tecnologia}")

# Ejercicio 4
total_tecnologia_funcional = sum(map(lambda v: v["cantidad"] * v["precio"], ventas_tecnologia))
print(f"Total acumulado de las ventas de tecnología: {total_tecnologia_funcional}")

# ANÁLISIS COMPARATIVO ENTRE EL ENFOQUE IMPERATIVO Y EL FUNCIONAL:
#    - El enfoque funcional aprovecha funciones de orden superior (map, filter)
#      y funciones anónimas (lambda) para encapsular la lógica en una sola
#      línea, haciendo el código más modular y libre de efectos secundarios.
```

```python
print("PARTE III – PROGRAMACIÓN ORIENTADA A OBJETOS")

# Definición de la clase Venta
class Venta:
    def __init__(self, producto, categoria, cantidad, precio):
        self.producto = producto
        self.categoria = categoria
        self.cantidad = cantidad
        self.precio = precio

    def obtener_monto(self):
        return self.cantidad * self.precio

# 1. Crear objetos
venta1 = Venta("Laptop", "Tecnologia", 2, 650)
venta2 = Venta("Mouse", "Tecnologia", 5, 20)
venta3 = Venta("Escritorio", "Muebles", 1, 300)
venta4 = Venta("Monitor", "Tecnologia", 2, 250)

# 2. Almacenar en una lista
lista_ventas_oop = [venta1, venta2, venta3, venta4]

# 3 Monto correspondiente a cada venta
print("Monto correspondiente a cada venta:")
for v in lista_ventas_oop:
    print(f"- Producto: {v.producto:<12} | Monto: {v.obtener_monto()}")

# 4. Determinar el monto total
monto_total_oop = sum(v.obtener_monto() for v in lista_ventas_oop)
print(f"\nMonto total de las ventas (POO): {monto_total_oop}")

# ANÁLISIS: DICCIONARIOS vs. OBJETOS (CLASES)
#    - Diccionario:Agrupa valores pero no tiene comportamiento propio.
#      Si queremos calcular el monto,
#      debemos escribir la lógica fuera de él.
#    - Objeto: Une los atributos con sus métodos.
#      La clase 'Venta' no solo almacena las propiedades sino que sabe cómo
#      calcular su propio monto a través del método 'obtener_monto()'.
```

```python
import time
from concurrent.futures import ThreadPoolExecutor

print("PARTE IV – PROGRAMACIÓN CONCURRENTE")

# Función que simula el procesamiento individual de una venta con un retraso de 1 segundo
def procesar_venta(venta):
    producto = venta["producto"] if isinstance(venta, dict) else venta.producto
    print(f"[Inicio] Procesando venta: {producto}...")
    time.sleep(1)
    monto = (venta["cantidad"] * venta["precio"]) if isinstance(venta, dict) else venta.obtener_monto()
    print(f"[Fin] Procesada venta: {producto} (Monto: {monto})")
    return monto

# Función para ejecutar el procesamiento bajo diferentes números de trabajadores
def ejecutar_pruebas_concurrencia(trabajadores):
    print(f"\n--- INICIANDO PRUEBA CON {trabajadores} TRABAJADOR(ES) ---")

    tiempo_inicio = time.time()

    with ThreadPoolExecutor(max_workers=trabajadores) as executor:
        resultados = list(executor.map(procesar_venta, ventas))

    tiempo_fin = time.time()
    tiempo_total = tiempo_fin - tiempo_inicio

    print(f"\nRESULTADOS:")
    print(f"- Cantidad de trabajadores: {trabajadores}")
    print(f"- Tiempo aproximado de ejecución: {tiempo_total:.2f} segundos")

    # Análisis del comportamiento observado según los resultados
    if trabajadores == 1:
        print("- Comportamiento observado: Las tareas se procesaron de forma estrictamente secuencial (una tras otra). Dado que cada venta tiene un retraso artificial de 1 segundo, el tiempo final es cercano a 6 segundos (6 ventas * 1s/venta).")
    elif trabajadores == 3:
        print("- Comportamiento observado: Las tareas se dividieron en lotes concurrentes de 3. Al procesar de 3 en 3, la carga total de 6 ventas se finaliza en 2 ciclos de procesamiento, resultando en un tiempo cercano a 2 segundos.")
    elif trabajadores == 6:
        print("- Comportamiento observado: Al contar con un trabajador dedicado para cada una de las 6 ventas, todas iniciaron y terminaron prácticamente al mismo tiempo, reduciendo el tiempo total de procesamiento a aproximadamente 1 segundo.")

# Ejecutar las tres pruebas
ejecutar_pruebas_concurrencia(1)
ejecutar_pruebas_concurrencia(3)
ejecutar_pruebas_concurrencia(6)
```
